# **Difference between `print(df.head())` and `df.head()`**

**`df.head()`**
* Returns a DataFrame (or Series view)
* In notebooks (Jupyter, Colab), it is automatically displayed
* It is meant for interactive use

**`print(df.head())`**
* Converts the DataFrame into a plain text string
* Prints it to the console/output stream
* Loses rich formatting (no nice table styling)

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
````

# Bar chart with some style

````python
# Data
array1 = [2, 7, 1, 9, 4]
array2 = [5, 3, 8, 6, 0]

# X positions
x = np.arange(len(array1))
width = 0.35

# Create figure
plt.figure(figsize=(8, 5))

# Bars
bars1 = plt.bar(x - width/2, array1, width=width, label="Example one")
bars2 = plt.bar(x + width/2, array2, width=width, label="Example two")

# Add value labels on top
for bar in bars1:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.1,
             f'{height}', ha='center', va='bottom')

for bar in bars2:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.1,
             f'{height}', ha='center', va='bottom')

# Labels and title
plt.xlabel('Bar Number')
plt.ylabel('Bar Height')
plt.title('Bar Graph Comparison')

# X ticks
plt.xticks(x, [f'Bar {i+1}' for i in x])

# Legend
plt.legend()

# Grid for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show plot
plt.tight_layout()
plt.show()
````

# **Barchart with dark_background style**

````python
import seaborn as sns

# Data
array1 = [2, 7, 1, 9, 4]
array2 = [5, 3, 8, 6, 0]

# Style
sns.set_theme(style="darkgrid")
plt.style.use("dark_background")

# X positions
x = np.arange(len(array1))
width = 0.35

# Figure
plt.figure(figsize=(9, 5))

# Bars with nicer colors
bars1 = plt.bar(x - width/2, array1, width=width, label="Example One")
bars2 = plt.bar(x + width/2, array2, width=width, label="Example Two")

# Add values on top
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, height + 0.2,
                 f'{height}', ha='center', va='bottom', fontsize=9)

# Labels and title
plt.xlabel('Bar Number', fontsize=11)
plt.ylabel('Bar Height', fontsize=11)
plt.title('📊 Bar Graph Comparison Dashboard', fontsize=14, weight='bold')

# X ticks
plt.xticks(x, [f'B{i+1}' for i in x])

# Legend
plt.legend()

# Grid styling
plt.grid(axis='y', linestyle='--', alpha=0.3)

# Remove top/right borders for clean look
sns.despine()

# Layout
plt.tight_layout()
plt.show()
````

# **Histogram Design**

````python
# 🔥 Reset any previous dark styles
#plt.rcdefaults()
# Data
nat = np.random.laplace(loc=15, scale=3, size=100)

# Histogram
counts, bins, patches = plt.hist(
    nat,
    bins='auto',
    alpha=0.9,
    rwidth=0.85,
    edgecolor='white'
)

# 🔥 Color each bar differently (gradient effect)
colors = plt.cm.viridis(np.linspace(0.2, 0.9, len(patches)))

for patch, color in zip(patches, colors):
    patch.set_facecolor(color)

# 🔥 Add values on top of bars
for count, patch in zip(counts, patches):
    plt.text(
        patch.get_x() + patch.get_width() / 2,
        count + 0.5,
        int(count),
        ha='center',
        va='bottom',
        fontsize=9,
        color='white'
    )

# Labels
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('📊 Enhanced Histogram Dashboard', fontsize=14, weight='bold')

# Grid
plt.grid(axis='y', alpha=0.3)

# Text annotation
plt.text(22, max(counts)*0.9, r'$\mu=15,\ b=3$', fontsize=11, color='white')

# Clean y-limit
maxfreq = max(counts)
plt.ylim(0, maxfreq + 5)

plt.tight_layout()
plt.show()
````

# **Load Weather Data**

````python
# Load data
weather = pd.read_csv(
    'https://raw.githubusercontent.com/alanjones2/dataviz/master/london2018.csv'
)
````

````python
# 🔥 Reset any previous dark styles
plt.rcdefaults()

# Create figure and axis
fig, ax = plt.subplots(figsize=(10, 5))

# 🔹 Force ash/gray background
fig.patch.set_facecolor('#eeeeee')
ax.set_facecolor('#eeeeee')

# Plot lines
ax.plot(weather['Month'], weather['Tmin'], marker='o', label='Tmin', linewidth=2)
ax.plot(weather['Month'], weather['Tmax'], marker='o', label='Tmax', linewidth=2)

# Value labels Tmin
for x, y in zip(weather['Month'], weather['Tmin']):
    ax.text(x, y - 0.5, str(y), ha='center', fontsize=8)

# Value labels Tmax
for x, y in zip(weather['Month'], weather['Tmax']):
    ax.text(x, y + 0.3, str(y), ha='center', fontsize=8)

# Labels
ax.set_xlabel("Month")
ax.set_ylabel("Temperature (°C)")
ax.set_title("London Weather Trend (Tmin vs Tmax)")

# Grid (soft gray)
ax.grid(True, linestyle='--', alpha=0.4)

# Legend
ax.legend()

plt.tight_layout()
plt.show()
````

# **Histogram plot example**

````python
np.random.seed(0)
height= np.round(np.random.normal(1.75, 0.20, 7000), 2)
weight= np.round(np.random.normal(60.32, 15, 7000), 2)
````

````python
# Style (clean + professional)
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))

# Histogram
counts, bins, patches = plt.hist(
    weight,
    bins=10,
    edgecolor='black',
    alpha=0.85,
    color='#4C72B0'
)

# 🔢 Add values on top of bars
for count, patch in zip(counts, patches):
    plt.text(
        patch.get_x() + patch.get_width()/2,
        count,
        int(count),
        ha='center',
        va='bottom',
        fontsize=9
    )

# Labels
plt.xlabel("Weight")
plt.ylabel("Frequency")
plt.title("Weight Distribution Histogram", fontsize=14, weight='bold')

# Grid
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
````

# **Boxplot**

````python
# Style
sns.set_theme(style="whitegrid")

plt.figure(figsize=(6, 5))

# Boxplot
plt.boxplot(
    height,
    patch_artist=True,
    boxprops=dict(facecolor="#4C72B0", color="black"),
    medianprops=dict(color="yellow", linewidth=2),
    whiskerprops=dict(color="black"),
    capprops=dict(color="black"),
    flierprops=dict(marker='o', markerfacecolor='red', markersize=6, alpha=0.6)
)

# Labels
plt.ylabel("Height")
plt.title("Height Distribution Boxplot", fontsize=14, weight='bold')

# Grid (y-axis only for clarity)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
````

````python
# Style
sns.set_theme(style="whitegrid")

plt.figure(figsize=(6, 5))

# Boxplot
plt.boxplot(
    weight,
    patch_artist=True,
    boxprops=dict(facecolor="#4C72B0", color="black"),
    medianprops=dict(color="yellow", linewidth=2),
    whiskerprops=dict(color="black"),
    capprops=dict(color="black"),
    flierprops=dict(marker='o', markerfacecolor='red', markersize=6, alpha=0.6)
)

# Calculate stats
q1 = np.percentile(weight, 25)
median = np.median(weight)
q3 = np.percentile(weight, 75)
min_val = np.min(weight)
max_val = np.max(weight)

# 🔢 Add annotations
plt.text(1.05, median, f"Median: {median:.2f}", va='center', fontsize=9)
plt.text(1.05, q1, f"Q1: {q1:.2f}", va='center', fontsize=9)
plt.text(1.05, q3, f"Q3: {q3:.2f}", va='center', fontsize=9)
plt.text(1.05, min_val, f"Min: {min_val:.2f}", va='center', fontsize=9)
plt.text(1.05, max_val, f"Max: {max_val:.2f}", va='center', fontsize=9)

# Labels
plt.ylabel("Weight")
plt.title("Weight Distribution Boxplot", fontsize=14, weight='bold')

# Grid
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
````

# **Interactive Boxplot (Plotly)**

````python
import plotly.graph_objects as go

# Create boxplot
fig = go.Figure()

fig.add_trace(go.Box(
    y=weight,
    name="Weight",
    boxpoints="all",   # show all data points
    jitter=0.3,        # spread points
    pointpos=-1.8,     # position of points
    marker=dict(color="rgba(76, 114, 176, 0.6)"),
    line=dict(color="rgb(76, 114, 176)"),
    hoverinfo="y"      # interactive hover values
))

# Layout styling
fig.update_layout(
    title="Interactive Weight Distribution (Boxplot)",
    yaxis_title="Weight",
    template="plotly_white",
    width=700,
    height=500
)

fig.show()
````
