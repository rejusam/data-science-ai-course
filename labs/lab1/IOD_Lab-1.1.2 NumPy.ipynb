## Summary statistics with height, weight and population data

````python
np.random.seed(0)
height= np.round(np.random.normal(1.75, 0.20, 7000), 2)
weight= np.round(np.random.normal(60.32, 15, 7000), 2)
population=np.column_stack((height, weight))
````

````python
print('height:\n',height)
print('weight:\n',weight)
print('population:\n',population)
````

**10. Create separate histograms of height and weight**

````python
help(plt.hist)
````

````python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.hist(height, bins=30, edgecolor='yellow')
plt.title('Height Distribution')
plt.xlabel('Height (m)')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(weight, bins=30, edgecolor='red')
plt.title('Weight Distribution')
plt.xlabel('Weight (kg)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
````

**11. Create separate boxplots of height and weight**

````python
help(plt.boxplot)
````

````python
plt.figure(figsize=(10, 6))
plt.subplot(1, 2, 1)
plt.boxplot(height)
plt.title('Height Boxplot')
plt.ylabel('Height (m)')

plt.subplot(1, 2, 2)
plt.boxplot(weight)
plt.title('Weight Boxplot')
plt.ylabel('Weight (kg)')

plt.tight_layout()
plt.show()
````

**12. Create a scatterplot of height and weight**

````python
help(plt.scatter)
````

````python
plt.figure(figsize=(8, 6))
plt.scatter(height, weight, alpha=0.5)
plt.title('Height vs Weight Scatter Plot')
plt.xlabel('Height (m)')
plt.ylabel('Weight (kg)')
plt.grid(True)
plt.show()
````
